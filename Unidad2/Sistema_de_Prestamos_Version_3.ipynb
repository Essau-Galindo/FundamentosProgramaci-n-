```python
from datetime import datetime

# ========================================
# SISTEMA DE PRÉSTAMOS
# Versión 1 - Programación basica y uso de listas
# ========================================

# Lista de alumnos activos en el sistema
alumnos = [
    ["26040200", "superman", "Ingenieria en Inteligencia Artificial"],
    ["26040201", "Bat Man", "Ingenieria Industrial"],
    ["26040202", "Wonder Woman", "Ingenieria en Inteligencia Artificial"],
    ["26040203", "Spider Man", "Ingenieria en Sistemas computacionales"],
    ["26040204", "Iron Man", "Ingenieria en Inteligencia Artificial"]
]

# Cada valor en la lista de computadoras tiene su lista correspondiente con el
# aula donde esta ubicada, el software instalado y el estado de esa computadora
computadoras = [
               ["PC-LC1-01","LC1","Python, VS Code","Disponible"],
               [ "PC-LC1-02","LC1","Python, Java, VS Code","Prestada"],
               ["PC-LC1-03","LC1","AutoCAD, MATLAB","Reservada"],
               [ "PC-LC2-01","LC2","Python, VS Code","Disponible"],
               ["PC-LC2-02","LC2","Python, Java, VS Code","Prestada"],
               ["PC-LC2-03","LC2","AutoCAD, MATLAB","Disponible"],
               ["PC-LC3-01", "LC3","Python, VS Code","Disponible"],
               ["PC-LC3-02","LC3","Python, Java, VS Code","Prestada"],
               ["PC-LC3-03","LC3","AutoCAD, MATLAB","Reservada"],
               ["PC-LC3-04","LC3", "Python, VS Code","Mantenimiento"],
            ]

historial_prestamos = []

necesidades = []

def listar_computadoras():
  print("\n--- COMPUTADORAS ---")
  for i in range(len(computadoras)):
      print(i+1, "Aula:", computadoras[i][1],"| ID:", computadoras[i][0],
            "| Software:", computadoras[i][2]," |Estado:", computadoras[i][3] )

def bucar_sw():
  programa = input( "\nIngrese el software que desea buscar: ")
  print("\n--- RESULTADOS ---")
  encontrado = False
  for i in range(len(computadoras)):
      # la funcion (lower) convierte a minuscula el texto, de esta forma
      # nos aseguramos que el texto a buscar y la informacion de las listas
      # este en minuscula y no tengamos problema de que no encunentra un programa
      # porque este escrito de forma diferente usando mayusculas y minusculas
      if programa.lower() in computadoras[i][2].lower():
          print(i,  "Aula:", computadoras[i][1],"ID:", computadoras[i][0],
                "Software:", computadoras[i][2],"Estado:", computadoras[i][3])

          encontrado = True
  if encontrado == False:
      print("No se encontraron computadoras con ese software.")

def solicitar_prestamo():
  print("\n===SOLICITAR PRESTAMO===")
  num_control= input("Numero de control del alumno: ")

  alumno_existe = False
  for i in range(len(alumnos)):
    if alumnos [i][0] == num_control:
        alumno_existe = True
        print("Alumno:", alumnos[i][1], "|", alumnos[i][2])
        break

  if alumno_existe == False:
      print("el alumno no existe en la lista")
      return
  programa = input("Que programa necesita: ")

  sw_encontrado = False   # ¿el programa existe en alguna computadora?
  asignada = False        # ¿se logró prestar una computadora?

  for i in range(len(computadoras)):
      if programa.lower() in computadoras[i][2].lower():
          sw_encontrado = True
          if computadoras[i][3] == "Disponible":
              computadoras[i][3] = "Prestada"
              print("Se asignó la computadora", computadoras[i][0],
                    "que esta en el aula", computadoras[i][1])
              registro = [num_control, programa, computadoras[i][1], computadoras[i][0]]
              historial_prestamos.append(registro)
              asignada = True
              break
          else:
              print("Si se encontro el software pero la", computadoras[i][0],
                    "esta", computadoras[i][3])

  # Si el programa no existe, preguntamos antes de agregarlo a necesidades
  if sw_encontrado == False:
      print("El programa", programa, "no existe en ninguna computadora.")
      respuesta = input("¿Desea agregarlo a la lista de necesidades? (s/n): ")
      if respuesta.lower() == "s":
          fecha_hora = datetime.now().strftime("%d/%m/%Y %H:%M:%S")
          necesidades.append([programa, fecha_hora])
          print("Programa agregado a la lista de necesidades.")
      else:
          print("No se agregó a la lista de necesidades.")
  elif asignada == False:
      print("No hay computadoras disponibles con ese software.")



def devolver_computadora():
  print("\n ===DEVOLVER COMPUTADORA====")
  id_pc = input("Dame el ID de la computadora")
  existe = False
  for i in range( len(computadoras) ):
    if computadoras[i][0] == id_pc:
      existe = True
      if computadoras[i][3]== "Prestada" or computadoras[i][3]=="Reservada":
        computadoras[i][3] = "Disponible"
        print("La computadora", id_pc, "ya esta disponible")
        break
      else:
        print("La computadora", id_pc, "no esta prestada/reservada")
  if existe == False:
    print("No existe la computadora", id_pc)


def agregar_computadora():
  print("\n--- AGREGAR COMPUTADORA ---")
  id_computadora = input("ID de la computadora: ")
  aula = input("Aula: ")
  programas = input("Software instalado: ")
  nueva_compu = [id_computadora, aula, programas, "Disponible"]
  computadoras.append(nueva_compu)
  print("\nComputadora agregada correctamente.")

def mostrar_historial():
  print("\n--- HISTORIAL DE PRESTAMOS ---")
  if len(historial_prestamos) == 0:
    print("No hay prestamos registrados.")
  else:
    for i in range(len(historial_prestamos)):
      print(i+1, "Alumno:", historial_prestamos[i][0],
            "| Programa:", historial_prestamos[i][1],
            "| Aula:", historial_prestamos[i][2],
            "| ID:", historial_prestamos[i][3])

def mostrar_necesidades():
  print("\n--- NECESIDADES DE SOFTWARE ---")
  if len(necesidades) == 0:
    print("No hay necesidades de software.")
  else:
    for i in range(len(necesidades)):
      print(i+1, "Programa:", necesidades[i][0],
            "| Fecha/Hora:", necesidades[i][1])

def mostrar_estadisticas():
  print("\n--- ESTADISTICAS ---")
  if len(historial_prestamos) == 0:
    print("No hay prestamos registrados.")
    return

  # --- Conteo por programa ---
  conteo_programas = []
  for i in range(len(historial_prestamos)):
      programa = historial_prestamos[i][1].lower()
      encontrado = False
      for j in range(len(conteo_programas)):
          if conteo_programas[j][0] == programa:
              conteo_programas[j][1] += 1
              encontrado = True

      if encontrado == False:
          conteo_programas.append([programa, 1])

  # --- Conteo por Carrera ---
  conteo_carreras = []
  for i in range(len(historial_prestamos)):
      num_control = historial_prestamos[i][0]
      carrera = ""
      # Buscar la carrera en la lista de alumnos
      for j in range(len(alumnos)):
          if alumnos[j][0] == num_control:
              carrera = alumnos[i][2]
              # Contar la carrera
              encontrado = False
              for k in range(len(conteo_carreras)):
                  if conteo_carreras[k][0] == carrera:
                      conteo_carreras[k][1] += 1
                      encontrado = True

              if encontrado == False:
                  conteo_carreras.append([carrera, 1])
                  break


      # --- Mostrar resultados por carrera---
      print("\n\n===================================")
      print("\nEstadisticas por carrera:")
      for i in range(len(conteo_carreras)):
          print(conteo_carreras[i][0],":", conteo_carreras[i][1], "veces")

  # --- Mostrar resultados por programa ---
  print("\n===================================")
  print("\n\nEstadisticas por programa:")
  for i in range(len(conteo_programas)):
      print(conteo_programas[i][0],":", conteo_programas[i][1], "veces")


# MENÚ PRINCIPAL
# Usamos un ciclo While que se repetira indefinidamente hasta que el usuario
# teclee la opcion 7 (salir) o cualquier numero superior a 7
opcion = 0
while opcion != 10:
    print("\n================================")
    print("      SISTEMA DE PRÉSTAMOS")
    print("================================")
    print("1. Consultar computadoras")
    print("2. Buscar por software")
    print("3. Solicitar préstamo")
    print("4. Devolver computadora")
    print("5. Crear reservación")
    print("6. Agregar computadora")
    print("7. Historial prestamos")
    print("8. Necesidades de software")
    print("9. Estadisticas")
    print("10. Salir")
    print("================================")
    opcion = int(input("Seleccione una opción: "))

    # 1. Consultar computadoras
    # si el usuario selecciono la opcion 1 se hace un ciclo (for) que recorre
    # toda la lista de computadoras e imprime los valores de las demás listas
    if opcion == 1:
        listar_computadoras()

    # 2. Buscar por software
    elif opcion == 2:
      bucar_sw()
    # 3. Solcitar prestamo
    elif opcion == 3:
      solicitar_prestamo()

    # 4. Devolver computadora
    elif opcion == 4:
      devolver_computadora()

    # 6. Agregar computadora
    elif opcion == 6:
      agregar_computadora()

    # 7. Historial prestamos
    elif opcion == 7:
      mostrar_historial()

    # 8. Necesidades de sotfware
    elif opcion == 8:
      mostrar_necesidades()

    # 9. Estadisticas
    elif opcion == 9:
      mostrar_estadisticas()

    # 10. salir
    elif opcion == 10:
        print("\nGracias por utilizar el sistema.")
        break
    else:
        print("\nOpción no válida. Intente de nuevo.")
```
