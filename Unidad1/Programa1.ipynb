```python
tabla = int(input("¿Que tabla de multiplicar quieres?"))
inicio = int(input("¿Desde donde quieres que empiece?"))
final = int(input("¿Hasta donde quieres que llegue?"))
incrementos = int(input("Dame el numero de los incrementos"))
# Agregamos la instruccion while para que valide que el valor
# inicial sea mayor que el final, de lo contraro me pedira
# indefinidamente el vaor final hasta que sea mayor
while inicio >= final:
    print("El limite final debe ser mayor que el inicio")
    final = int(input("Dame el limite final"))

for i in range(inicio,final,incrementos):
    print(tabla, "*", i, "=", tabla * i)
```

#**Listas en python**

```python
# Podemos acceder a cada elemento o item de la lista a traves
# del indice o su posicion. Las listas siempre empiezan
alumnos = ["Ana", "Luis", "Carlos"]
print(alumnos)
print(alumnos[0])
print(alumnos[2])
```

```python
# Podemos modificar el valor de un elemnto de la lista
# Usando el nombre de la lista y el indice y le asignamos
# el nuevo valor
print(alumnos)
alumnos[1] = "Juan"
print(alumnos)
```

```python
# Podemos especificar indices negativos y cuando se ponen valores
# negativos se lee o se accede a la lista desde el final
# empezamos con -1
print(alumnos)
print(alumnos[-1])
print(alumnos[-2])
```

```python
# El metodo append agrega un elemento al final de la lista
print(alumnos)
alumnos.append("Sandra")
print(alumnos)
```

```python
print(alumnos)
alumnos.insert(2, "Sofia")
print(alumnos)
```

```python
# Remove borra un elemnto de la lista especificando el
# valor, si existen valores repetidos, borra el primer
# valor que encuentra
print(alumnos)
alumnos.remove("Sofia")
print(alumnos)
```

```python
print(alumnos[2])
# la siguuiente linea marca error porque queremos acceder
# a un indice que no existe. La lista tiene unicamente 6 elemntos y
# queremos acceder al elemnto 7 que no existe
print(alumnos[7])
```

```python
# Pop me permite borrar un elemnto de la lista
# especificando su posicion
print(alumnos)
alumnos.pop(2)
print(alumnos)
```

```python
# Pop me permite borrar un elemnto de la lista
# especificando su posicion
print(alumnos)
del alumnos [2]
print(alumnos)
```

```python
# la funcion len me permite conocer el numero de
# elementos en la lista
print("Numero de elementos o en la lista", len(alumnos))
```

```python
# Ciclo para recorrer la lista y hacer algun proceso
alumnos.append("Sandra")
alumnos.append("Carlos")
print(alumnos)
for x in alumnos:
  print(x)
  if x == "Sofia":
    print("Encontramos a sofia")
    break # break forza a salir de un ciclo
```
